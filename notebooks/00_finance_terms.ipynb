## Before §1 — Words needed to understand trading venues

- **Asset:** something of economic value that a person or organization can own, such as a share or a building.
- **Security:** a financial instrument such as a share or bond that represents an ownership interest or a claim for payment.
- **Buy order / sell order:** an instruction stating what someone wants to buy or sell, how much, and sometimes at what price.
- **Bid:** a price a buyer is willing to pay. **Ask (offer):** a price a seller is willing to accept.
- **Matching orders:** finding a compatible buyer and seller. For example, a buyer willing to pay $51$ and a seller willing to accept $50$ can trade, subject to the venue's rules.
- **Execution:** the point at which an order results in an agreed trade.
- **Clearing:** the process after a trade of confirming and organizing who owes what to whom; it may involve a clearing organization that manages obligations and risk.
- **Settlement:** the actual transfer of money and securities to complete the trade. A trade can be executed now and settled later.
- **Listing:** having a security admitted to trading on a particular exchange under its rules.
- **Broker:** a firm or person that arranges trades for a customer. A broker is not the same thing as the marketplace.

**Example:** You ask a broker to buy one share. A compatible seller is found (matching), the trade is agreed (execution), obligations are processed (clearing), and payment and share ownership are transferred (settlement).

## 1 — Markets: local exchanges, OTC, and ECNs

**Exchange.** An organized marketplace with rules about which securities may trade, how orders interact, and how trades are reported and completed. Many exchanges are electronic; a physical trading floor is not required. 

**Example:** New York Stock Exchange. 

**Non-example:** A private agreement to sell a used bicycle is not a securities exchange.

**Over-the-counter (OTC) market.** Trading arranged outside a centralized exchange, often through dealers or directly between institutions. OTC does *not* necessarily mean informal or unregulated. 

**Example:** Two institutions negotiate a customized interest-rate swap. 

**Non-example:** Buying a listed share through an exchange order book is not, by itself, an OTC negotiation.

**Electronic Communication Network (ECN).** An electronic system that displays or matches orders from participants, following defined rules. Unlike a dealer, an ECN need not buy securities for its own inventory. 

**Example:** An electronic system that connects a buyer's and seller's compatible orders. 

**Non-example:** A financial news website showing prices but not arranging trades.

**Key distinction:** Exchange and OTC describe *market arrangements*; an ECN describes an *electronic order-matching system*. [1, 2, 3]

## Before §2 — Words needed for currencies and money

- **Currency:** the unit of money used in an economy, such as the US dollar (USD) or Indian rupee (INR).
- **Foreign exchange (FX):** exchanging one currency for another.
- **Exchange rate:** how much of one currency buys one unit of another.
- **Cash:** money immediately available for payment; it differs from an investment in a company.
- **Deposit:** money placed in a bank account, usually creating a claim against the bank.

**Example:** If 1 USD = 85 INR, exchanging 10 USD yields 850 INR before fees. 

**Non-example:** A company's share price is not an exchange rate. [1, 4]

## 2 — Products: money and FX currencies

**Money** serves as a medium of exchange, a unit in which prices are quoted, and a way to store purchasing power (imperfectly, because inflation can reduce it). **FX currency trading** exchanges money denominated in different currencies.

An **FX pair**, such as USD/INR, specifies a rate between two currencies. Currency movements matter to travelers, importers, exporters, and international investors. 

**Example:** An importer needing USD to pay a supplier is exposed to a change in USD/INR. 

**Non-example:** Buying shares of a US company is an equity investment, not directly an FX trade, although a foreign investor may also face currency risk. [1, 4]

## Before §3 — Words needed for stocks and funds

- **Company / corporation:** an organization that can own assets, owe debts, and issue ownership shares.
- **Equity:** an ownership interest in a company; it is different from a loan to the company.
- **Share (stock):** one unit of that ownership.
- **Dividend:** money a company may distribute to its shareholders; it is not guaranteed.
- **Index:** a numerical measure constructed to summarize a group of assets according to a rule.
- **Fund:** a pooled investment arrangement: many investors contribute money that is invested according to a stated policy.
- **Portfolio:** the collection of investments held by a person or fund.
- **ETF:** short for *exchange-traded fund*; its shares trade on an exchange.


### Example: Apple Inc. — From a Company to an ETF

Consider **Apple Inc. (AAPL)**, the company that designs and sells products such as the iPhone, MacBook, and iPad.

1. **Company / Corporation:** Apple Inc. is a publicly traded corporation. It owns assets such as offices, equipment, intellectual property, and cash. It also has financial obligations.

2. **Equity:** Suppose you purchase shares of Apple. You now own a very small fraction of the company. This ownership interest is called *equity*. You are an owner, not a lender to Apple.

3. **Share (Stock):** Apple's ownership is divided into billions of shares. If you purchase 10 shares of Apple, you own 10 units of its stock. These shares can generally be bought and sold on the stock market.

4. **Dividend:** Apple has historically paid cash dividends to shareholders. For example, if Apple declares a dividend of $0.25 per share, an investor holding 10 shares receives:

   $$
   10 \times \$0.25 = \$2.50.
   $$

   Dividends are not guaranteed; the company decides whether and how much to distribute.

5. **Index:** Apple is one of the companies included in the **S&P 500 index**, which tracks the performance of approximately 500 leading publicly traded US companies. An index is a numerical measure, not a company or an investment fund.

6. **Fund:** Instead of buying Apple shares individually, you could invest in a fund that holds shares of many companies, including Apple. The fund pools investors' money and follows a particular investment strategy.

7. **Portfolio:** Suppose you own 10 Apple shares, 5 Microsoft shares, and some US government bonds. Together, these investments form your *portfolio*.

8. **ETF (Exchange-Traded Fund):** The **Vanguard S&P 500 ETF (VOO)** is an ETF designed to track the S&P 500 index. By purchasing shares of VOO, you gain exposure to a portfolio containing Apple and hundreds of other companies, rather than purchasing every company's shares yourself.

**Important distinction:**

- **Apple (AAPL):** A company whose shares you can own.
- **S&P 500:** An index measuring the performance of a group of companies.
- **VOO:** An investment fund whose shares trade on a stock exchange and which seeks to track the S&P 500.
- **Your portfolio:** The collection of investments you personally hold.

## 3 — Stocks, equity indices, and ETFs

**Stock / equity:** A share represents partial ownership of a company, with rights determined by the share class and law. Its price can rise or fall, and dividends are not assured.

**Equity index:** A calculated measure of the performance of a selected group of stocks, for example the S&P 500. Different indices use different rules for selecting and weighting companies.

**Exchange-traded fund (ETF):** A fund whose shares can be bought and sold on an exchange. An ETF may track an index, but not every ETF does. An ETF share represents an interest in the fund's portfolio, not direct ownership of every underlying share.

**Example:** Buying an index-tracking ETF can give exposure to many companies through one fund share. 

**Non-example:** Buying one company's stock is not automatically a diversified fund investment. [1, 4, 5]

## Before §4 — Words needed for issuance

- **Raise capital:** obtain money for an activity or business.
- **Issue shares:** create and sell securities to investors.
- **Issuer:** the organization that creates the security.
- **Public offering:** an offering of securities to the public under applicable rules.
- **IPO:** *initial public offering*, a company's first public offering of its shares.
- **Proceeds:** money received from a sale.

**Example:** When a company sells newly issued shares, the company generally receives the proceeds. When an existing investor sells shares to another investor, the seller receives the proceeds. [1, 4]

## 4 — IPOs, primary markets, and secondary markets

**Primary market:** The market in which newly issued securities are sold and the issuer receives the proceeds. 

**Secondary market:** Trading of securities that have already been issued, usually between investors.

**IPO:** The first time a company offers its shares to the public; an IPO may include newly issued shares, existing shares, or both, so not every dollar in an IPO necessarily goes to the company.

**Example:** A company issues 100 new shares for $10$ each: it raises $1,000$ before expenses.

Later, Alice sells 10 of those shares to Bob for $12$ each: Alice receives $120$; the company does not receive that resale money.

**Non-example:** A resale of an existing share is not a new issuance. [1, 4]


### Example: Primary Market vs. Secondary Market

Consider **Apple Inc. (AAPL)**, a publicly traded company.

#### 1. Primary Market: A Company Raises Capital

Suppose Apple wants to raise money to build a new manufacturing facility.

To do this, Apple decides to issue **1 million new shares** at a price of **$200 per share**.

Investors purchase these newly issued shares, and Apple receives:

$$
1{,}000{,}000 \times \$200 = \$200{,}000{,}000.
$$

Thus, Apple raises **$200 million** (before transaction costs).

This transaction takes place in the **primary market**, because the company is selling newly issued securities to investors.

**Key observations:**

- Apple issues new shares.
- Investors pay money to Apple (possibly through financial intermediaries).
- Apple receives capital that it can use for its business.
- The total number of outstanding Apple shares increases.

An **Initial Public Offering (IPO)** is a common example of a primary-market transaction. However, an already publicly traded company like Apple can also issue additional shares through a *follow-on offering*.

#### 2. Secondary Market: Investors Trade Existing Shares

Now suppose Alice owns **10 shares of Apple**, which she purchased earlier.

She decides to sell them at **$210 per share**.

Bob purchases all 10 shares from Alice through a stock exchange.

The transaction value is:

$$
10 \times \$210 = \$2{,}100.
$$

This transaction takes place in the **secondary market**, because Alice is selling shares that already exist.

**Key observations:**

- Alice receives $2,100$ (before transaction costs).
- Bob becomes the new owner of the 10 shares.
- Apple does **not** receive the $2,100$.
- The total number of outstanding Apple shares does not change.

Most everyday stock trading takes place in the secondary market.

#### 3. Comparison

| Feature | Primary Market | Secondary Market |
|---|---|---|
| What is traded? | Newly issued securities | Existing securities |
| Who sells? | The issuing company | An existing investor |
| Who receives the money? | The issuing company | The selling investor |
| Does the company raise new capital? | Yes | No |
| Example | Apple issues new shares | Alice sells Apple shares to Bob |

**Important principle:** The primary market helps companies raise capital, while the secondary market allows investors to buy and sell existing investments.

The secondary market also provides **liquidity**: investors have an opportunity to convert their investments into cash by selling them to other investors.

Liquidity does not guarantee that an investor can sell immediately or at the desired price.

### Question: If Bob buys Apple shares from Alice, will he receive dividends from Apple?

**Answer: Yes!** Bob will be entitled to receive dividends from Apple after purchasing Apple shares from Alice, provided he satisfies the eligibility requirements.

This illustrates an important principle:

**Dividends belong to the eligible shareholder, not necessarily the person who originally purchased the shares from the company.**

#### Example: Alice sells Apple shares to Bob

Suppose:

1. Alice owns 100 shares of Apple.
2. Apple declares a dividend of $0.30$ per share.
3. Alice sells all 100 shares to Bob for $200$ per share.
4. Bob now owns those 100 Apple shares.

If Bob purchases the shares before the *ex-dividend date* and satisfies the applicable trading rules, he will receive:

$$
D = 100 \times \$0.30 = \boxed{\$30}.
$$

Apple pays this dividend even though Bob purchased the shares from Alice rather than directly from Apple.

#### The Ex-Dividend Date

The **ex-dividend date** determines whether a buyer is entitled to an upcoming dividend.

Suppose Apple announces the following hypothetical schedule:

| Event | Date |
|---|---|
| Dividend declaration | March 1 |
| Ex-dividend date | March 10 |
| Dividend payment | March 20 |

- **Bob buys on March 9:** Bob is generally entitled to the upcoming dividend.
- **Bob buys on March 10:** Alice generally receives the upcoming dividend, even though Bob now owns the shares.
- **Future dividends:** Bob becomes eligible for subsequent dividends as long as he continues to hold the shares through the relevant eligibility dates.

This assumes ordinary stock trades under standard US settlement rules.

#### Why Does Apple Pay Bob?

A share represents an **ownership interest** in a company.

When Alice sells her shares to Bob, she transfers that ownership interest and the associated shareholder rights.

Therefore:

- Alice is no longer the owner of those shares.
- Bob becomes the new shareholder.
- Bob acquires the rights associated with those shares, including eligibility for future declared dividends.
- Apple does not need to issue new shares or receive new money for Bob to become a shareholder.

This is a fundamental feature of the **secondary market**: ownership changes hands, but the rights attached to the securities continue.

#### Important Distinction: Dividends vs. Bond Interest

- **Stock dividends:** Payments made to shareholders when a company declares a dividend. Common-stock dividends are not guaranteed.
- **Bond interest:** Payments that a bond issuer is contractually obligated to make according to the bond's terms, subject to the issuer's ability to pay.

Thus, owning Apple shares does not guarantee that Bob will receive dividends. However, if Apple declares a dividend and Bob meets the eligibility requirements, he is entitled to receive it.

#### Key Takeaway

**Buying shares in the secondary market transfers ownership rights, including eligibility for future dividends, from the seller to the buyer.**

The ex-dividend date determines which investor receives a particular declared dividend.


## Before §5 — Words needed for lending and interest rates

- **Borrower:** receives money now and promises repayment under agreed terms.
- **Lender / creditor:** supplies money and holds a claim for repayment.
- **Principal:** the amount borrowed, apart from interest.
- **Interest:** the cost of borrowing money, commonly expressed as a percentage over time.
- **Interest rate:** the rule or percentage used to calculate interest.
- **Maturity:** the date a loan or bond is due to be repaid under its terms.
- **Default:** failure to meet a required payment or another important contractual obligation.
- **Credit risk:** the possibility that a borrower will not pay as promised.

**Example:** Lending $100$ at 5% simple annual interest for one year gives a promised repayment of $105$. 

**Non-example:** A stockholder does not normally have a contractual right to repayment of the share purchase price. [1, 4]

## 5 — Loans, bonds, credit products, and interest rates

**Loan:** An agreement under which a lender provides money to a borrower, who promises to repay according to specified terms.

**Bond:** A transferable debt security: an issuer borrows from investors and promises payments specified in the bond contract. Bonds may pay periodic **coupons** (interest payments) and repay a **face value** (stated principal) at maturity. Some bonds pay no coupons.

**Credit product:** A financial arrangement involving a promise to pay, or exposure to the risk that such a promise is broken. Loans and bonds are familiar examples; some derivatives also transfer credit risk.

**Interest rate:** A price for borrowing over time; its precise interpretation depends on compounding, term, and contract details.

**Example:** A government issues bonds to finance spending. 

**Non-example:** A shareholder's ownership interest is not a bond. The pricing mathematics is reserved for Notebook 01. [1, 4]

## Before §6 — Words needed for commodities

- **Physical good:** something tangible, such as wheat, copper, or oil.
- **Standardization:** defining common specifications so buyers know what quantity and quality they are trading.
- **Spot transaction:** an agreement for near-term purchase and delivery, in contrast with a contract specifying a later date.
- **Supply and demand:** how much sellers offer and buyers wish to purchase at different prices.

### Example: Standardization and Spot Transactions in Commodity Markets

Consider a company that manufactures electrical wires. It needs **copper**, an important commodity used in electrical equipment.

#### 1. Standardization: What exactly are we buying?

Suppose the company wants to purchase **10 tonnes of copper**.

However, not all copper is identical:

- Some copper is 99.99% pure.
- Some copper is only 95% pure.
- Copper may be supplied in different shapes and sizes.
- Different suppliers may use different measurement systems.

How can buyers and sellers agree on exactly what is being traded?

This is where **standardization** becomes important.

**Definition:** Standardization means establishing common specifications for a commodity, such as its quality, purity, quantity, and acceptable delivery conditions.

For example, a standardized copper contract might specify:

| Property | Specification (illustrative) |
|---|---|
| Commodity | Copper |
| Purity | At least 99.99% |
| Quantity | 10 tonnes |
| Form | Copper cathodes |
| Delivery location | A specified warehouse |

Now suppose Alice wants to sell copper and Bob wants to buy copper.

If both agree to trade copper meeting these specifications, they do not need to negotiate every physical characteristic separately.

**Why is standardization useful?**

1. It makes it easier to compare prices from different sellers.
2. It reduces uncertainty about the quality and quantity being purchased.
3. It allows commodities from different suppliers to be traded under common specifications.
4. It makes organized commodity markets more efficient.

**Important:** Standardization does not mean that all copper in the world is identical. It means that copper satisfying a particular set of specifications can be traded under common rules.

#### 2. Spot Transaction: Buying Copper Now

Suppose Bob's electrical-wire company needs copper for immediate production.

Alice offers to sell **10 tonnes of copper** at a price of **$9,000 per tonne**.

Bob agrees to purchase it at today's market price.

The total transaction value is:

$$
P = 10 \times \$9,000 = \boxed{\$90,000}.
$$

Alice and Bob agree to exchange the copper and payment promptly, according to the market's settlement and delivery arrangements.

This is called a **spot transaction**.

**Definition:** A spot transaction is an agreement to buy or sell an asset at the current market price, with payment and delivery occurring promptly, usually according to a short, established settlement schedule.

The price agreed upon is called the **spot price**.

**Important:** A spot transaction does not necessarily mean that physical delivery occurs instantly. Depending on the commodity and market, delivery may take some time.

#### 3. Spot Transaction vs. Forward Contract

Suppose Bob does not need copper today. Instead, he needs it **three months from now**.

Alice and Bob agree today that:

- Alice will deliver 10 tonnes of copper in three months.
- Bob will pay \$9,200 per tonne upon delivery.
- The agreed total payment will be \$92,000.

This is a **forward contract**, not a spot transaction.

| Feature | Spot Transaction | Forward Contract |
|---|---|---|
| When is the price agreed upon? | Today | Today |
| When does delivery occur? | Promptly | At a specified future date |
| Price | Current spot price | Price agreed today for future delivery |
| Example | Buy copper for near-term delivery | Agree today to buy copper in three months |

Notice that **both transactions agree on a price today**. The crucial difference is when the commodity is scheduled to be delivered.

#### 4. How Are Standardization and Spot Transactions Related?

These concepts describe different aspects of a trade:

- **Standardization** tells us *what exactly is being traded*.
- **Spot transaction** tells us *when the trade is to be settled and delivered*.

A spot transaction can involve a standardized commodity, but it does not have to. For example, two businesses can negotiate a spot purchase of copper with customized specifications.

Similarly, standardized commodities can be traded through spot transactions or contracts for future delivery.

## Before §7 — Words needed for real estate and securitization

- **Real estate:** land and buildings.
- **Mortgage:** a loan secured by property; if the borrower does not pay, the lender may have legal rights over the property.
- **Collateral:** an asset pledged as security for a debt.
- **Cash flow:** money paid or received over time.
- **Pool:** a collection of many assets or contracts.
- **Securitization:** arranging assets such as loans into securities whose payments depend on the underlying assets.
- **ABS:** *asset-backed security*, a security backed by cash flows from a pool of assets.


### Example: Alice Buys a House — Real Estate, Mortgages, and Securitization

Let us consider a hypothetical scenario in which **Alice wants to purchase a house**. We will use this example to understand real estate, mortgages, collateral, cash flows, pooling, securitization, and asset-backed securities (ABS).

#### 1. Real Estate: Alice Purchases a House

**Real estate** refers to land and buildings, including residential houses, apartments, offices, and commercial properties.

Suppose Alice wants to purchase a house worth **$500,000**.

However, Alice has only $100,000 in savings. She needs to borrow the remaining amount.

#### 2. Mortgage: Alice Borrows Money from a Bank

A **mortgage** is a loan secured by real estate.

Alice approaches a bank and agrees to the following arrangement:

| Item | Amount |
|---|---|
| House price | $500,000 |
| Alice's savings (down payment) | $100,000 |
| Mortgage loan | $400,000 |
| Annual interest rate | 6% (fixed) |
| Loan duration | 30 years |

The bank lends Alice $400,000, which she uses along with her savings to purchase the house.

Alice agrees to repay the loan through monthly payments over 30 years.

For a conventional fully amortizing mortgage with monthly payments, the monthly payment is approximately **$2,398**.

Each monthly payment contains two components:

- **Principal:** The portion that reduces the amount Alice owes.
- **Interest:** The amount the bank charges Alice for borrowing money.

Initially, a larger portion of Alice's payment goes toward interest. Over time, as the outstanding loan balance decreases, a larger portion goes toward principal.

#### 3. Collateral: What Happens if Alice Cannot Repay?

**Collateral** is an asset pledged as security for a loan.

In Alice's case, the house serves as collateral for her mortgage.

Suppose Alice stops making mortgage payments.

The bank may have the legal right to initiate **foreclosure**, a process through which the lender can recover the property and sell it to recover the unpaid debt, subject to applicable laws.

For example:

- Alice's outstanding mortgage balance is \$350,000.
- Alice stops making payments.
- Following the required legal procedures, the property is sold for \$370,000.

The proceeds may be used to repay the outstanding loan and applicable costs. Any remaining amount is handled according to the relevant laws and contractual terms.

**Important:** Collateral reduces the lender's potential loss, but does not eliminate risk. If the house sells for less than the outstanding debt and associated costs, the lender may still suffer a loss.

#### 4. Cash Flow: Alice Makes Monthly Payments

**Cash flow** refers to money moving into or out of a person, company, or investment over time.

In our example:

- Alice receives \$400,000 from the bank when the mortgage is originated.
- The bank provides \$400,000 as a loan.
- Alice subsequently makes monthly mortgage payments of approximately \$2,398.
- The bank receives those payments as cash inflows.

From Alice's perspective, the monthly payments are **cash outflows**.

From the bank's perspective, they are **cash inflows**.

The mortgage therefore generates a stream of expected future payments.

This is important because those future payments have financial value.

#### 5. Pool: Combining Many Mortgages

Suppose the bank has issued mortgages to many customers:

| Borrower | Outstanding Mortgage |
|---|---:|
| Alice | \$400,000 |
| Bob | \$300,000 |
| Charlie | \$500,000 |
| Diana | \$250,000 |
| Emma | \$350,000 |

Together, these five mortgages have an outstanding principal of:

$$
400{,}000+300{,}000+500{,}000+250{,}000+350{,}000
=\boxed{\$1{,}800{,}000}.
$$

A **pool** is a collection of financial assets or contracts grouped together.

Instead of considering each mortgage separately, the bank or another financial institution may combine them into a mortgage pool.

Each borrower continues making payments on their respective mortgage.

The combined payments form the **cash flows of the mortgage pool**.

In practice, mortgage pools may contain hundreds or thousands of loans.

#### 6. Securitization: Turning Mortgages into Investable Securities

Now suppose the bank wants to obtain money today rather than wait many years to collect all the mortgage payments.

One possibility is **securitization**.

**Securitization** is the process of pooling financial assets, such as mortgages, and creating securities whose payments depend on the cash flows generated by those assets.

A simplified securitization process works as follows:

1. The bank originates mortgages to Alice, Bob, Charlie, Diana, and Emma.
2. The mortgages are transferred to a separate legal entity, often called a **Special Purpose Vehicle (SPV)**.
3. The SPV finances the purchase of the mortgages by issuing securities to investors.
4. Investors pay money to purchase these securities.
5. Borrowers continue making their monthly mortgage payments.
6. The mortgage payments are collected and used to pay investors, after applicable fees and according to the securities' contractual terms.

**Why would a bank do this?**

By selling mortgages through securitization, the bank can obtain funds earlier rather than waiting for borrowers to repay their loans over many years.

This can free up funding for additional lending and transfer some mortgage-related risks to investors.

However, securitization does not eliminate the underlying risks.

#### 7. Asset-Backed Securities (ABS): What Do Investors Purchase?

An **asset-backed security (ABS)** is a financial security whose payments depend on cash flows generated by a collection of underlying assets.

In our example, the underlying assets are mortgages.

A security backed specifically by mortgages is commonly called a **Mortgage-Backed Security (MBS)**, a type of asset-backed security in the broad sense.

Suppose an investor named Frank purchases a security backed by the mortgage pool containing Alice's loan.

Frank does not own Alice's house directly.

Instead, Frank owns a financial security that entitles him to payments determined by the terms of that security.

Those payments are supported by the cash flows generated by the mortgages.

For example:

- Alice makes her monthly mortgage payment.
- Other borrowers make their payments.
- The payments are collected.
- After fees and according to the contractual payment rules, money is distributed to investors such as Frank.

**What happens if Alice stops paying?**

The mortgage pool receives less money than expected from Alice's loan.

This may affect the payments received by investors, depending on the structure of the security, protections against losses, and the performance of other mortgages in the pool.

Thus, investors in mortgage-backed securities are exposed to risks associated with the underlying mortgages.

#### 8. Putting Everything Together

| Financial Term | Example from Alice's Story |
|---|---|
| **Real estate** | Alice's \$500,000 house |
| **Mortgage** | The \$400,000 loan Alice obtains from the bank |
| **Collateral** | Alice's house, which secures the mortgage |
| **Cash flow** | Alice's monthly mortgage payments |
| **Pool** | A collection of mortgages issued to Alice and other borrowers |
| **Securitization** | The process of transferring and packaging mortgages into securities sold to investors |
| **ABS** | A security backed by cash flows from a pool of assets |
| **MBS** | A security specifically backed by mortgages |

#### Important Distinction

There are three different assets or financial claims in this example:

1. **The house:** A physical real-estate asset owned by Alice.
2. **The mortgage:** A loan representing Alice's obligation to repay borrowed money.
3. **The mortgage-backed security:** A financial instrument purchased by investors whose payments depend on the mortgage pool.

These should not be confused.

In particular, **buying a mortgage-backed security does not mean buying the houses themselves**.

## 7 — Real estate, mortgages, and ABS

**Real estate** refers to land and buildings and can be held for use, rent, or potential resale.

**Mortgage:** A loan secured by real estate. The borrower generally makes payments to the lender, and the property acts as collateral.

**Asset-backed security (ABS):** A security whose promised payments are supported by cash flows from a pool of financial assets such as loans. Mortgage-backed securities are a related, more specific class backed by mortgages.

## Before §8 — Words needed for derivatives

- **Contract:** an agreement establishing rights and obligations.
- **Underlying:** the asset, price, rate, or event on which a contract's value depends.
- **Notional amount:** a reference quantity used to calculate payments; it is not always exchanged.
- **Payoff:** the payment or economic result specified by a contract, often at a future date.
- **Hedge:** an arrangement intended to reduce exposure to an unwanted price or rate change.
- **Leverage:** gaining large economic exposure relative to the money initially committed; this can amplify gains and losses.
- **Counterparty:** the other party to a contract.


## 8 — Options, futures, forwards, swaps, derivatives, and structured products

**Derivative:** A financial contract whose value or payments depend on an underlying asset, price, rate, or event.

**Forward:** A customized agreement today to buy or sell something at a specified future time and price; typically negotiated privately.

**Future:** A standardized, exchange-traded contract for a specified future transaction, with exchange and clearing arrangements.

**Option:** A contract giving its buyer the **right, but not the obligation**, to buy (call) or sell (put) an underlying asset on specified terms. The seller has a corresponding obligation if the buyer exercises the right.

**Swap:** A contract in which two parties exchange specified cash flows, for example fixed-rate payments against floating-rate payments.

**Structured product:** A financial instrument combining features of multiple instruments, often to create a particular pattern of potential payments and risks.


### Example: Understanding Derivatives Through Hypothetical Financial Transactions

We will use several hypothetical scenarios involving farmers, companies, banks, and investors to understand the terminology associated with **derivatives**.

The central idea is that a derivative is a financial contract whose value depends on something else, such as the price of wheat, the price of a stock, or an interest rate.

---

#### 1. Contract: An Agreement Between Two Parties

Suppose Alice is a wheat farmer and Bob owns a bakery.

Alice expects to harvest wheat in six months. Bob needs wheat to produce bread.

They agree today that:

- Alice will sell Bob 10 tonnes of wheat.
- Bob will pay $300 per tonne.
- The wheat will be delivered six months from now.

The total agreed payment is:

$$
10 \times \$300 = \$3,000.
$$

This agreement is a **contract** because it establishes rights and obligations for both parties.

- **Alice's obligation:** Deliver 10 tonnes of wheat.
- **Bob's obligation:** Pay $3,000 according to the agreed terms.

A contract specifies what each party must do and under what conditions.

#### 2. Underlying: What Determines the Contract's Value?

In the previous example, the contract concerns wheat.

Therefore, **wheat is the underlying commodity**.

More precisely, the market price of wheat influences the economic value of the contract.

Suppose the market price of wheat rises to $400 per tonne after six months.

Bob still has the contractual right to purchase wheat for $300 per tonne.

Compared with purchasing wheat at the new market price, Bob saves:

$$
10(400-300)=\$1,000.
$$

Thus, the value of Bob's contractual position depends on the market price of wheat.

Other examples of underlyings include:

- Apple stock prices.
- Gold prices.
- Foreign exchange rates.
- Interest rates.
- Stock market indices such as the S&P 500.

#### 3. Notional Amount: The Reference Value of a Contract

Suppose Alice and Bob enter into a financial agreement involving 100 tonnes of wheat at a reference price of $300 per tonne.

The **notional amount** is:

$$
N=100\times \$300=\boxed{\$30,000}.
$$

This amount provides a reference for calculating contractual payments.

However, the notional amount is **not necessarily exchanged** between the parties.

For example, suppose their contract is settled entirely in cash.

If wheat rises to $350 per tonne, Alice might pay Bob:

$$
100(350-300)=\$5,000.
$$

Here, the actual cash payment is $\$5,000$, not $\$30,000$.

**Important distinction:**

- **Notional amount:** Reference amount used to calculate payments.
- **Actual payment:** Money transferred according to the contract's terms.

In some derivatives, such as physically delivered forwards, the full purchase price may be exchanged. In others, such as many swaps, only payments calculated from the notional are exchanged.

#### 4. Payoff: What Does the Contract Pay?

A **payoff** is the payment or economic result specified by a financial contract, often determined at its expiration.

Return to Alice and Bob's wheat contract.

Suppose Bob agrees to purchase 10 tonnes of wheat for $300 per tonne in six months.

Let:

- $S_T$ = market price of wheat per tonne after six months.
- $K=300$ = agreed purchase price per tonne.
- $Q=10$ = quantity of wheat.

The payoff to Bob's forward position at maturity is:

$$
\Pi_T=Q(S_T-K).
$$

Consider three possibilities:

| Wheat price at maturity | Bob's payoff |
|---|---:|
| $\$250$ per tonne | $-\$500$ |
| $\$300$ per tonne | $\$0$ |
| $\$400$ per tonne | $+\$1,000$ |

For example, when $S_T=400$:

$$
\Pi_T=10(400-300)=\boxed{\$1,000}.
$$

Bob benefits because he can purchase wheat below its prevailing market price.

If wheat falls to $\$250$, Bob has a negative payoff relative to buying at the market price, because he is still obligated to pay \$300.

**Note:** This payoff measures the economic value of the forward position at maturity. It is not necessarily Bob's overall business profit, since his bakery also uses the wheat to produce bread.

#### 5. Hedge: Reducing Exposure to Risk

Suppose Bob owns a bakery and worries that wheat prices will increase over the next six months.

If wheat becomes more expensive, producing bread will cost more.

Bob enters into a forward contract to purchase wheat at $300 per tonne.

Now consider two scenarios:

**Scenario A: Wheat prices rise to $400.**

Without the contract, Bob would pay:

$$
10\times400=\$4,000.
$$

With the contract, he pays:

$$
10\times300=\$3,000.
$$

He avoids the additional $1,000 expense.

**Scenario B: Wheat prices fall to $250.**

Without the contract, Bob could purchase wheat for:

$$
10\times250=\$2,500.
$$

But his contract requires him to pay $3,000.

Bob loses the opportunity to benefit from the lower price.

This arrangement is called a **hedge**.

**Definition:** Hedging means taking a financial position intended to reduce exposure to an unwanted risk.

The goal of hedging is not necessarily to maximize profit. It is to make future financial outcomes more predictable.

#### 6. Leverage: Controlling a Large Position with Less Initial Capital

Suppose Charlie wants exposure to the price of gold.

One possibility is to purchase $100,000 worth of gold outright.

However, Charlie instead enters into a futures contract representing $100,000 worth of gold.

Assume the exchange requires Charlie to deposit $10,000 as **initial margin**.

Margin is money deposited as financial security to support the contract and cover potential losses.

Charlie has exposure to $\$100,000$ worth of gold while initially committing $\$10,000$.

The ratio is:

$$
L=\frac{\text{Notional Exposure}}{\text{Initial Margin}}
=\frac{100,000}{10,000}
=\boxed{10}.
$$

We describe this as approximately **10 times leverage** relative to the initial margin.

Suppose gold prices increase by 5%.

The approximate gain on Charlie's position is:

$$
0.05\times100,000=\$5,000.
$$

Relative to his initial $\$10,000$ margin, this is:

$$
\frac{5,000}{10,000}=50\%.
$$

However, if gold prices fall by 5%, Charlie loses approximately $\$5,000$, or 50% of his initial margin.

**Important:** Leverage amplifies both gains and losses. Charlie may also need to deposit additional money if his account balance falls below the required margin level.

#### 7. Counterparty: Who Is on the Other Side?

Return to Alice and Bob's wheat contract.

- Alice promises to deliver wheat.
- Bob promises to pay for the wheat.

From Alice's perspective, Bob is her **counterparty**.

From Bob's perspective, Alice is his **counterparty**.

Now suppose wheat prices rise sharply, and Alice refuses or is unable to deliver the wheat at the agreed price.

Bob may suffer a financial loss because his counterparty failed to fulfill the contract.

This introduces **counterparty risk**.

**Definition:** Counterparty risk is the possibility that the other party to a financial contract fails to fulfill its obligations.

Counterparty risk is particularly important in privately negotiated financial contracts.

---

### 8. Forward Contract: Agreeing Today on a Future Transaction

The wheat contract between Alice and Bob is an example of a **forward contract**.

A forward is an agreement made today to buy or sell something at a specified future date for an agreed price.

For example:

- Underlying: Wheat.
- Quantity: 10 tonnes.
- Forward price: $300 per tonne.
- Delivery date: Six months from today.

Alice and Bob can negotiate the quantity, delivery date, quality, and other conditions.

This makes forwards **customizable**.

Forward contracts are commonly traded privately, rather than through an organized exchange.

They also expose the parties to counterparty risk.

### 9. Futures Contract: A Standardized Forward-Like Agreement

Suppose Bob does not want to negotiate a private contract with Alice.

Instead, he enters into a wheat futures contract on an organized commodity exchange.

The exchange specifies standardized terms, such as:

- The quantity of wheat represented by one contract.
- Acceptable grades of wheat.
- Contract expiration dates.
- Delivery or cash-settlement arrangements.

Bob chooses an available contract rather than negotiating all these conditions individually.

Unlike a typical forward contract, a futures contract generally involves:

1. **Initial margin:** Money deposited to support the position.
2. **Daily settlement:** Gains and losses are calculated and credited or debited as prices change.
3. **Clearinghouse:** An institution that stands between buyers and sellers through the clearing process and manages important aspects of counterparty risk.

For example, suppose Bob enters into a futures position representing 10 tonnes of wheat at $300 per tonne.

If the futures settlement price rises to $310 per tonne the next day, his long position gains:

$$
10(310-300)=\$100.
$$

This gain is credited through the futures settlement process.

**Main distinction:**

- **Forward:** Typically privately negotiated and customizable.
- **Future:** Standardized, exchange-traded, and subject to margin and daily settlement rules.

Both can be used to hedge against future price changes.

### 10. Option: The Right, but Not the Obligation

Suppose David believes Apple stock will increase in price.

Apple shares currently trade at $200 each.

David purchases a **call option** giving him the right to buy one Apple share for $210 in three months.

The agreed purchase price of $210 is called the **strike price**.

Suppose David pays $8 for this option. This initial payment is called the **option premium**.

**Scenario A: Apple rises to $250.**

David can exercise his option and purchase a share for $\$210$ when its market value is $250.

The payoff is:

$$
\max(250-210,0)=\$40.
$$

After subtracting the $8 premium, his net profit is:

$$
40-8=\boxed{\$32}.
$$

**Scenario B: Apple falls to $180.**

David has no reason to buy a share for $\$210$ when he can purchase it in the market for $\$180$.

He allows the option to expire.

His option payoff is zero, and his net loss is the $8 premium.

For a call option with strike price $K$, the payoff at expiration is:

$$
\Pi_T=\max(S_T-K,0).
$$

A **put option**, by contrast, gives its buyer the right, but not the obligation, to sell the underlying asset at the strike price.

Its payoff at expiration is:

$$
\Pi_T=\max(K-S_T,0).
$$

**Important:** The option buyer has a right, while the option seller has a corresponding obligation if the option is exercised.

The example assumes a hypothetical option covering one share. Standard US listed equity option contracts typically represent 100 shares.

### 11. Swap: Exchanging Cash Flows

Suppose Alice owns a company that has borrowed $1 million from a bank.

Her loan has a **floating interest rate**, meaning the rate changes periodically with a specified market reference rate.

Alice worries that interest rates will increase.

She enters into an **interest-rate swap** with another financial institution.

Under the swap:

- Alice agrees to pay a fixed annual rate of 5%.
- The other institution agrees to pay a floating rate.
- Both payments are calculated using a notional principal of $1 million.

Suppose the floating rate for a particular one-year period is 7%.

Alice's fixed payment under the swap is:

$$
0.05\times1,000,000=\$50,000.
$$

The floating payment she receives is:

$$
0.07\times1,000,000=\$70,000.
$$

Her net receipt from the swap is:

$$
70,000-50,000=\boxed{\$20,000}.
$$

If the floating rate on her loan is also 7% over that period, she pays $\$70,000$ in loan interest but receives $\$20,000$ net from the swap.

Her combined interest expense is:

$$
70,000-20,000=\boxed{\$50,000}.
$$

Thus, the swap has effectively converted her floating-rate borrowing cost into a fixed-rate cost of 5%, under the simplifying assumption that the reference rates and payment dates match.

**Important:** The $1 million notional principal is generally not exchanged in a plain-vanilla interest-rate swap. It is used to calculate the interest payments.

### 12. Structured Product: Combining Financial Instruments

Suppose Emma wants to invest in the stock market but is concerned about losing her initial investment.

A bank offers her a hypothetical three-year structured note with the following terms:

- Initial investment: $10,000.
- Underlying: S&P 500 index.
- If the index rises, Emma receives 80% of its positive percentage return.
- If the index falls, Emma receives her original $10,000 at maturity, provided the issuing bank fulfills its obligations.
- No additional interest payments.

**Scenario A: The index rises by 20%.**

Emma receives:

$$
10,000\left(1+0.80\times0.20\right)
=\boxed{\$11,600}.
$$

**Scenario B: The index falls by 20%.**

Under the stated terms, Emma receives:

$$
\boxed{\$10,000}.
$$

The product can be understood conceptually as combining:

1. A debt instrument intended to provide a specified repayment at maturity.
2. An option-like component providing exposure to stock market gains.

This is an example of a **structured product**.

**Important risks:**

- The issuing bank may default.
- Emma may not recover her initial investment if she sells the note before maturity.
- She may receive less upside than an investor who directly owns the underlying stocks.
- The product may be difficult to sell before maturity.

Therefore, *principal protection* does not mean that the investment is completely risk-free.

### 13. Comparison of Derivative Instruments

| Instrument | Main Purpose | Defining Feature |
|---|---|---|
| **Forward** | Lock in a future transaction price | Privately negotiated future agreement |
| **Future** | Hedge or speculate on future prices | Standardized, exchange-traded contract |
| **Call option** | Benefit from rising prices or hedge certain risks | Right to buy at a specified price |
| **Put option** | Benefit from falling prices or protect against losses | Right to sell at a specified price |
| **Swap** | Exchange financial cash flows | Parties exchange payments according to agreed rules |
| **Structured product** | Create a customized risk-return profile | Combines multiple financial features |

### Answer to Exercise 1

#### 1. How many different securities does the investor hold?

**Answer: 3 different securities.**

The investor owns:

- 20 shares of Apple (AAPL).
- 10 shares of Microsoft (MSFT).
- 5 shares of Vanguard S&P 500 ETF (VOO).

A **security** is a tradable financial instrument representing, for example, ownership in a company or a financial claim.

Although the investor owns $20+10+5=35$ shares in total, these represent only **three distinct securities**.

#### 2. Is VOO a company, an index, or a fund?

**Answer: VOO is an Exchange-Traded Fund (ETF).**

VOO is managed by Vanguard and seeks to track the performance of the S&P 500 index.

The distinction is:

- **Apple:** A company.
- **S&P 500:** An index measuring the performance of a collection of companies.
- **VOO:** A fund that invests in companies represented in the S&P 500.

An investor cannot directly purchase the S&P 500 index itself, but can purchase shares of an ETF such as VOO that tracks it.

#### 3. Does the investor have exposure to Apple through more than one investment?

**Answer: Yes, through two investments.**

The investor has exposure to Apple in two ways:

1. **Direct exposure:** The investor owns 20 Apple shares.
2. **Indirect exposure:** The investor owns 5 shares of VOO, which holds Apple shares as part of its underlying portfolio.

This introduces an important concept called **overlapping exposure**.

For example, suppose Apple represents 7% of VOO's portfolio (a hypothetical weight). Then approximately 7% of the value invested in VOO is also exposed to Apple's performance.

Consequently, owning Apple shares and VOO does not provide as much diversification away from Apple as one might initially think.

#### 4. If Apple pays a dividend of $0.30 per share, how much does the investor receive directly?

**Answer: $6.00.**

The dividend payment is calculated by multiplying the number of shares by the dividend per share:

$$
D = N \times d
$$

where:

- $N=20$ is the number of Apple shares.
- $d=\$0.30$ is the dividend per share.

Therefore,

$$
D=20\times \$0.30=\boxed{\$6.00}.
$$

The investor receives $6 directly from the Apple shares, assuming they are eligible for that dividend.

**What about the 5 VOO shares?**

The investor may also receive distributions from VOO. However, those are separate fund distributions, determined by the ETF's holdings and distribution policy. They are not calculated simply by multiplying the number of VOO shares by Apple's dividend.

#### Additional observation: Diversification and exposure

This exercise illustrates three important concepts in portfolio management:

- **Number of securities:** How many distinct financial instruments are held.
- **Diversification:** How investments are spread across different assets and sources of risk.
- **Exposure:** How much the portfolio's value is affected by a particular company, industry, or market.

In particular, **owning more securities does not necessarily mean having a more diversified portfolio**, because different securities can contain overlapping investments.



### Answer to Exercise 2

#### 1. Apple issues $500 million worth of new bonds to investors. Is this a primary-market or secondary-market transaction?

**Answer: Primary market.**

Apple is issuing new bonds to raise capital from investors. Since these bonds are being sold for the first time, the transaction takes place in the **primary market**.

#### 2. Alice sells an Apple bond she already owns to Bob. Does Apple receive any money from this transaction?

**Answer: No.**

This is a **secondary-market transaction** because Alice is selling an existing bond to Bob.

- Bob pays Alice for the bond.
- Alice receives the money.
- Apple does not receive any new capital from this transaction.

Apple's contractual obligations under the bond remain in place, but Bob becomes the new bondholder.

#### 3. Why might investors be less willing to purchase newly issued shares if there were no secondary market?

**Answer: Because they would have difficulty selling their shares in the future.**

This is a problem of **liquidity**.

Liquidity refers to how easily an asset can be converted into cash without significantly affecting its price.

Suppose Alice buys Apple shares for $2,000. A year later, she needs money for an emergency.

- **With a secondary market:** Alice can attempt to sell her shares to another investor.
- **Without a secondary market:** Alice may struggle to find someone willing to buy her shares, even if Apple is performing well.

Therefore, investors may hesitate to buy newly issued shares if they cannot easily sell them later.

**Important distinction:** Uncertainty about whether an investment will gain or lose value is called *investment risk*. A secondary market does not eliminate this uncertainty. Instead, it provides opportunities to sell investments.

#### 4. Can a company participate in both primary-market and secondary-market transactions?

**Answer: Yes, but its role is different in each market.**

Consider Apple:

- **Primary market:** Apple issues new shares or bonds to raise capital.
- **Secondary market:** Investors buy and sell existing Apple shares or bonds among themselves.

Apple is the **issuer** of the securities in both cases, but it generally receives new capital only when issuing securities in the primary market.

Apple may also participate directly in the secondary market, for example, by repurchasing its own shares through a *share buyback program*.

---

#### Key takeaway

The two markets serve complementary purposes:

- **Primary market:** Enables companies and governments to raise capital.
- **Secondary market:** Provides liquidity and allows investors to transfer ownership of existing securities.

A functioning secondary market can make investors more willing to participate in the primary market.



### Answers to Exercise 3

#### 1. Alice sells Bob 5 tonnes of standardized copper for prompt delivery. Is this a spot transaction?

**Answer: Yes.**

This is a **spot transaction** because Alice and Bob agree to exchange copper for prompt delivery.

The fact that the copper is standardized means that its quality, quantity, and other specifications are clearly defined.

However, standardization is not what makes this a spot transaction. The important feature is **prompt settlement and delivery**.

#### 2. Alice and Bob agree today to exchange 5 tonnes of copper six months later. Is this a spot transaction?

**Answer: No.**

This is a **forward contract**, assuming they agree today on the terms of the future exchange.

The important distinction is:

- **Spot transaction:** The commodity is purchased for prompt delivery.
- **Forward contract:** The parties agree today to exchange the commodity at a specified future date.

Both transactions can involve a price agreed upon today, but their delivery schedules differ.

#### 3. If two copper suppliers meet the same standardized specifications, must they charge the same price?

**Answer: No.**

Standardization ensures that the copper satisfies common specifications, but it does not require all suppliers to charge identical prices.

For example, suppose:

- Alice sells standardized copper for \$9,000 per tonne.
- Charlie sells copper meeting the same specifications for \$9,050 per tonne.

Both suppliers satisfy the same quality requirements, but their prices differ.

Possible reasons include transportation costs, delivery locations, availability, and differences in commercial terms.

**Important principle:** Standardization makes products easier to compare, but it does not impose a uniform price.

#### 4. Can a commodity be traded in the spot market without being standardized?

**Correct answer: Yes.**

A commodity does not need to be standardized to be traded in the spot market.

For example, suppose a farmer has harvested 500 kg of apples of varying sizes and quality.

A local grocery-store owner inspects the apples and agrees to purchase the entire batch for $600, with payment and delivery occurring immediately.

This is a **spot transaction**, even though the apples do not conform to a particular standardized grade.

The buyer and seller simply negotiate the quality, quantity, and price themselves.

**Important distinction:**

- **Standardization** concerns the specifications of the product being traded.
- **Spot trading** concerns the timing of payment and delivery.

These are independent characteristics of a transaction.

A commodity can therefore be:

1. Standardized and traded in the spot market.
2. Non-standardized and traded in the spot market.
3. Standardized and traded through a forward contract.
4. Non-standardized and traded through a forward contract.

---

#### Key Takeaway

**Standardization answers "What exactly are we trading?" while spot trading answers "When will the exchange take place?"**

A commodity does not need to be standardized to be traded in the spot market, and a standardized commodity does not necessarily have to be traded for immediate delivery.



### Answers to Exercise 4

#### 1. Alice purchases a $\$600,000$ house using $\$150,000$ of her savings. How much does she need to borrow from the bank?

**Answer: $450,000.**

The amount Alice needs to borrow is the difference between the house price and her down payment.

$$
\text{Mortgage} = \text{House Price} - \text{Down Payment}
$$

Therefore,

$$
600{,}000 - 150{,}000 = \boxed{\$450{,}000}.
$$

Alice contributes $\$150,000$ of her own money and borrows the remaining $\$450,000$ through a mortgage.

#### 2. If Alice fails to repay her mortgage, why might the bank have the right to sell her house?

**Answer: Because the house serves as collateral for the mortgage.**

When Alice takes out a mortgage, she agrees that the house will serve as **collateral**, meaning that the property secures her obligation to repay the loan.

If Alice fails to make the required payments, the bank may initiate a legal process called **foreclosure**.

Foreclosure allows the lender, subject to applicable laws and procedures, to recover and sell the property to help repay the outstanding debt.

**Important:** The bank does not automatically become the owner of the house when Alice misses a payment. Foreclosure requires following the applicable legal procedures.

#### 3. Suppose a bank combines 1,000 mortgages into a single pool. Does this mean that all borrowers now have one common mortgage?

**Answer: No.**

Each borrower still has a separate mortgage agreement with its own:

- Outstanding loan amount.
- Interest rate.
- Repayment schedule.
- Collateral property.

A **mortgage pool** is simply a collection of individual mortgage loans grouped together for financial purposes.

For example, Alice and Bob may have mortgages of $\$400,000$ and $\$300,000$, respectively.

Even if their mortgages belong to the same pool, Alice is not responsible for repaying Bob's mortgage, and Bob is not responsible for repaying Alice's mortgage.

The purpose of pooling is to combine the cash flows generated by many loans.

#### 4. Why might a bank prefer to securitize mortgages instead of keeping them until maturity?

**Answer: To obtain funds earlier and potentially transfer some financial risks to investors.**

Suppose a bank issues a 30-year mortgage to Alice.

If the bank keeps the mortgage, it receives principal and interest payments gradually over the life of the loan.

However, the bank may prefer to receive money sooner.

Through **securitization**, the bank can sell mortgages into a structure that issues securities to investors.

This can provide several benefits:

1. **Earlier access to funds:** The bank receives money from selling the mortgages rather than waiting for all future payments.
2. **Additional lending capacity:** The funds received can help finance new loans.
3. **Risk transfer:** Some risks associated with the mortgages may be transferred to investors.
4. **Balance-sheet management:** Depending on the transaction structure and accounting rules, the bank may reduce the amount of mortgage assets held on its balance sheet.

**Important:** Securitization does not eliminate mortgage risk. It redistributes some of that risk among financial institutions and investors.

#### 5. Frank purchases a mortgage-backed security containing Alice's mortgage. Does Frank now own Alice's house?

**Answer: No.**

Frank owns a **mortgage-backed security (MBS)**, not Alice's physical property.

The distinction is:

- **Alice:** Owns the house, subject to the mortgage.
- **The mortgage lender or current loan holder:** Holds the contractual claim to mortgage payments and associated security rights.
- **Frank:** Owns a financial security whose payments depend on cash flows from a pool of mortgages.

Frank may receive payments supported by Alice's mortgage payments, but he does not directly own Alice's house or gain the right to live in it.

#### 6. If several borrowers in the mortgage pool stop making payments, how might this affect investors holding the mortgage-backed securities?

**Answer: Investors may receive less money than expected or suffer financial losses.**

The payments to investors depend, directly or indirectly, on the cash flows generated by the underlying mortgages.

Suppose a mortgage pool contains 1,000 loans.

If all borrowers make their scheduled payments, the pool receives the expected cash flows.

However, if 100 borrowers stop making payments:

1. The pool receives less money than expected.
2. Some mortgages may enter **default**, meaning that borrowers fail to meet their contractual payment obligations.
3. The lender or loan servicer may attempt to recover money through foreclosure or other procedures.
4. If the recovered amounts are insufficient, the mortgage pool may experience losses.
5. Depending on the structure of the securities, investors may receive reduced payments or suffer losses.

Not every missed payment immediately causes an investor loss. Some securities have protections, such as reserves, guarantees, or arrangements that allocate losses differently among investors.

**Important principle:** Pooling mortgages can reduce dependence on any single borrower, but it does not eliminate the possibility of losses.

---

#### Key Takeaway

This exercise illustrates how a single real-estate transaction can become part of a much larger financial system:

$$
\text{House Purchase}
\longrightarrow
\text{Mortgage}
\longrightarrow
\text{Mortgage Pool}
\longrightarrow
\text{Securitization}
\longrightarrow
\text{Mortgage-Backed Securities}.
$$

The essential distinction is between **ownership of a physical asset**, **a contractual loan obligation**, and **a financial security backed by loan payments**.

Securitization allows financial institutions to transform collections of loans into securities that investors can purchase, while distributing the associated cash flows and risks according to the securities' terms.
