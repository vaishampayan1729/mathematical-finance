## 1. Two ways of thinking about a stock's value

Suppose Alice considers buying a share of a company. Two different questions can guide her:

1. **Firm-foundation approach:** *What is the share worth based on the cash flows it is expected to generate?*
2. **Castle-in-the-air approach:** *What price will someone else be willing to pay for the share later?*

These are not equivalent questions. A stock can be attractive to a short-term speculator even if its current price seems high relative to a fundamental valuation, provided the speculator expects other investors to pay an even higher price.

**Terminology.** The *market price* is the price at which the asset trades. An estimate of *intrinsic (fundamental) value* attempts to assess the asset from expected future economic benefits. Neither quantity is known with certainty in advance.

## 2. The castle-in-the-air theory

**Definition (informal).** The castle-in-the-air theory emphasizes the role of investor psychology and expectations about *future market opinion* in determining asset prices. An investor may buy an asset chiefly because they expect another investor to pay more for it later.

The phrase *castle in the air* suggests a valuation supported largely by expectations and enthusiasm rather than a firm estimate of underlying cash flows.

**Example: Alice buys a fashionable technology stock**

- The stock trades today at **$\$100$**.
- Alice's own cash-flow-based estimate of its value is only **$\$80$**.
- Alice nevertheless buys because she expects market excitement to push the stock to **$\$130$** next month.
- If she sells at $\$130$, her gross gain is $\$30$ per share, or **30%**, despite having believed the stock was already expensive on fundamental grounds.

But if enthusiasm fades and the stock falls to $\$70$, Alice loses $\$30$ per share. Her expected resale price was a prediction, **not a guarantee**.

**Non-example:** Alice buys a stock at $\$100$ because she estimates that its future dividends and other economic benefits justify a present value of $140. This reasoning is primarily firm-foundation reasoning, even if she also expects its market price to rise.

## 3. Keynes's newspaper beauty contest: the rules

John Maynard Keynes used a newspaper competition to explain the psychology of professional investment. Malkiel retells the analogy in his discussion of castle-in-the-air theory.

**The contest:**

1. A newspaper displays **100 photographs** of faces.
2. Each participant chooses **six** faces.
3. The winner is the participant whose choices most closely match the **group's overall preferences**.

**Crucial distinction:** The contest does **not** reward the participant for selecting the faces they personally find most attractive. It rewards the participant for anticipating **which faces other participants will choose**.

So the relevant question changes from:

> Which six faces do I prefer?

to:

> Which six faces do I think the other contestants will select?

The beauty contest is an analogy about *coordination of expectations*, not a claim that attractiveness has an objectively correct ranking.

**Source:** Keynes, *The General Theory of Employment, Interest and Money* (1936), Chapter 12, section V; Malkiel, *A Random Walk Down Wall Street*, Chapter 1.

## 4. Why the contest becomes complicated: expectations about expectations

Suppose the newspaper publishes photographs labeled A through Z (imagine that this is a smaller version of the original contest).

**Level 0 — Personal preference.** Alice chooses A, B, C because *she* likes them best. This ignores how the prize is awarded.

**Level 1 — Predict other people's preferences.** Bob chooses D, E, F because he thinks those are the most popular faces, even though he personally prefers A, B, C.

**Level 2 — Predict what others predict.** Carol thinks many contestants will reason like Bob. She therefore asks what *those contestants think the crowd will choose*, rather than asking directly what the crowd likes.

**Level 3 and beyond — Higher-order beliefs.** David reasons about what others think still other contestants will predict. This process can continue through further levels.

The striking feature is that the winning selection may emerge from **shared expectations about popularity**, not merely from the sum of sincere personal preferences.

**Important nuance:** Higher-order reasoning is not automatically more accurate. It depends on how other contestants actually behave. There is no simple guarantee that a player using more levels will win.

## 5. A numerical miniature beauty contest

For simplicity, replace 100 faces and six choices with **four faces and one choice**. Suppose the winner is whoever chooses the face selected by the largest number of participants.

| Face | Alice's personal preference score | Alice's prediction of the number of other votes |
|---|---:|---:|
| A | 10 | 2 |
| B | 8 | 12 |
| C | 7 | 5 |
| D | 6 | 1 |

Alice personally prefers **A**, but she predicts **B** will win the popularity contest.

- Choosing A follows her personal preference.
- Choosing B follows her prediction of the crowd.

If the prize depends on matching the crowd, **B is the rational choice given Alice's predictions**. But if her predictions are wrong, she can still lose.

## 6. Mapping the beauty contest to stock markets

| Newspaper contest | Financial market |
|---|---|
| Face in a photograph | Stock or other traded asset |
| Personal judgment of beauty | Investor's estimate of fundamental value |
| Guessing the crowd's selection | Forecasting future investor demand and sentiment |
| Winning by matching group choices | Profiting if future buyers value the asset more highly |
| Higher-order guesses | Expectations about other investors' expectations |

**Example: a popular AI company**

- Alice estimates a stock's fundamental value at **$\$70$**.
- Its market price is **$\$100$**.
- Alice expects other investors to become even more enthusiastic and bid the price up to **$\$125$**.
- Bob knows investors like Alice are making this bet, and he buys because he expects the enthusiasm to persist.

Alice and Bob need not believe the stock's underlying business is worth $\$125$. They may be betting on the *future market valuation*.

The strategy can succeed temporarily, but it is vulnerable to a change in sentiment. When investors stop expecting further buyers to pay higher prices, demand can reverse sharply.

## 7. Why Keynes's example matters

Keynes's analogy identifies a difference between:

- **Forecasting an asset's long-run economic benefits**, and
- **Forecasting the market's short-run valuation of those benefits**.

Even if Alice thinks a stock's long-run cash flows justify a price of $\$150$, she may avoid buying at $\$100$ if she expects the market price to fall to $80 before she plans to sell. Conversely, she might buy an asset she considers overvalued if she expects the market to value it even more highly soon.

**This does not mean fundamentals never matter.** Rather, short-run market prices can also depend on collective expectations, narratives, liquidity, and investor sentiment.

**Key idea:** In the beauty-contest setting, an investor's prediction is about **other people's predictions**, not just about the underlying asset.

## 8. Firm foundation versus castle in the air: worked comparison

Consider a share with an expected dividend of **$\$5$ next year**, growing at a constant rate of **3%** annually. Assume the appropriate discount rate is **8%**, with $r>g$.

A simplified constant-growth dividend discount model gives

$$
V_0=\frac{D_1}{r-g}=\frac{5}{0.08-0.03}=\boxed{\$100}.
$$

This is a **firm-foundation estimate**, subject to the model's strong assumptions.

Now suppose the share trades at **$\$130$**. Alice expects that market excitement will push it to **$\$160$** in one month.

- **Firm-foundation question:** Does the $\$130$ price make sense given the discounted future dividends? In this model, it is above the estimated $\$100$ value.
- **Castle-in-the-air question:** Will future buyers pay $\$160$ despite the $\$130$ current price?

The same stock can therefore look **expensive under one perspective** but **attractive to a speculator under the other**.

**Caution:** The $100 estimate is not an objectively certain value. Changing assumptions about dividends, growth, or discount rates changes the valuation.

## 9. Price, resale expectations, and risk

For a share bought at price $P_0$ and sold at $P_1$, ignoring dividends and costs, the realized return is

$$
R=\frac{P_1-P_0}{P_0}.
$$

If Alice buys at $\$100$, consider three possible resale prices:

| Future sale price | Gross return |
|---:|---:|
| $\$130$ | +30% |
| $\$100$ | 0% |
| $\$60$ | −40% |

The castle-in-the-air strategy depends critically on *when* sentiment changes and whether a willing buyer is available at the expected price.

This is closely related to the **greater-fool mechanism**: buying mainly because someone else might pay more later. Not every expectation-driven trade is irrational, but reliance on future enthusiasm can create significant downside risk.

## 10. A small Python experiment: resale prices

We can calculate returns for several hypothetical future prices. The calculation is deterministic: the uncertainty lies in which future price actually occurs.

```python
# Import NumPy to work with a collection of hypothetical future prices.
import numpy as np

# Set the current price paid by Alice.
purchase_price = 100.0

# Specify several possible prices at which Alice might later sell.
future_prices = np.array([60.0, 80.0, 100.0, 130.0, 160.0])

# Compute the gross return for each possible future sale price.
returns = (future_prices - purchase_price) / purchase_price

# Display the price and return for every scenario.
for price, result in zip(future_prices, returns):
    # Print each scenario with its corresponding percentage return.
    print(f"Future price: ${price:6.2f} | Gross return: {result:+.1%}")
```

## 11. What the theory does **not** claim

- It does **not** claim that prices are *always* detached from fundamentals.
- It does **not** claim that investors can reliably predict crowd psychology.
- It does **not** imply that every rising stock is a speculative bubble.
- It does **not** prove that markets are always irrational.
- It does **not** guarantee profits from buying a popular asset early.

**Non-example:** A company's stock price rises after an unexpectedly strong and sustainable increase in its future earnings. Such a price change may be explained by revised fundamental valuations, even if investor sentiment also plays a role.

## 13. Solutions

## 14. Summary

- **Firm foundation:** Focus on an asset's expected economic benefits and a valuation derived from them.
- **Castle in the air:** Focus on the price future market participants may be willing to pay.
- **Keynes's beauty contest:** To win, predict the crowd's choices—not simply your own preferences.
- **Higher-order expectations:** Investors may try to anticipate what others think still others will do.
- **Main risk:** Expectations can reverse, and a hoped-for future buyer may not appear at the anticipated price.

**Connection to the course:** These perspectives motivate why market prices cannot always be understood from discounted cash flows alone, and why expectations and uncertainty matter in mathematical finance.
